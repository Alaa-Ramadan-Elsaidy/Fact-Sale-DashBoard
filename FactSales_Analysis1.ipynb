## 2) Clean Data
before cleaning we knew that :

1- no duplication 

2- invoice date , delivery date is string (we will make it datetime)

3- delivery date has 13 nulls (we will fill it **delivery=invoice + 1**)لاحظت ان التوصيل يبقي بعد يوم واحد فقط في جميع الصفوف

```python
class Cleaner:
    def __init__(self, data):
        self.data = data.copy()

    def fix_columns_trim(self):
        self.data.columns = ( self.data.columns.str.strip() )
        print("trimmed successfully")
        return self

    def fix_dates(self, cols):
        for col in cols:
                self.data[col] = pd.to_datetime(self.data[col])
        print('we fixed dates successfully from string to date_time')
        return self

    def fill_delivery_date(self):
        self.data['Delivery Date Key'] = self.data['Delivery Date Key'].fillna( 
            self.data['Invoice Date Key'] + pd.Timedelta(days=1)
        )
        print("Delivery dates filled 13 rows successfully")
        return self

    def get_data(self):
        return self.data


cleaner = Cleaner(df)
df = (
    cleaner.fix_columns_trim()
    .fix_dates(["Invoice Date Key", "Delivery Date Key"])
    .fill_delivery_date()
    .get_data()
)
```

```python
def get_column_types(df):
    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
    datetime_cols = df.select_dtypes(include=['datetime', 'datetime64']).columns.tolist()
    categorical_cols = df.select_dtypes(include=['object', 'str']).columns.tolist()

    print("Numeric Columns:")
    print(numeric_cols)

    print("\nCategorical Columns:")
    print(categorical_cols)

    print("\ndate Columns:")
    print(datetime_cols)
    
```

### Outliers using IQR

```python
df.head(2)
```

```python
def flag_outliers(data, col):
    q1 = data[col].quantile(0.25)
    q3 = data[col].quantile(0.75)
    iqr = q3 - q1
    lower_fence = q1 - 1.5 * iqr
    upper_fence = q3 + 1.5 * iqr
    is_outlier = ~ data[col].between(lower_fence, upper_fence)
    return is_outlier


for col in ["Profit", "Tax Amount", "Total Including Tax", "Unit Price",'Quantity', 'Unit Price',]:
        df[col + "_outlier"] = flag_outliers(df, col)
        print(f"number of outliers in {col+'_outliers'}:", df[col + "_outlier"].sum())
```

## 3) Data Validation

```python
import pandas as pd

def data_validation(df: pd.DataFrame) -> pd.DataFrame:
    print('#' * 50)
    print('Validation Report:')
    print('#' * 50)

    # 1) Tax Amount Check: Tax = Total Excluding Tax * (Tax Rate / 100)
    calc_tax = df['Total Excluding Tax'] * df['Tax Rate'] / 100.0
    tax_diff = (calc_tax - df['Tax Amount']).abs()
    print(f"1) Tax Amount check -> records with diff > 0.01: {(tax_diff > 0.01).sum()}")

    # 2) Total Including Tax Check: Total = Total Excluding Tax + Tax Amount
    calc_total = df['Total Excluding Tax'] + df['Tax Amount']
    total_diff = (calc_total - df['Total Including Tax']).abs()
    print(f"2) Total Including Tax check -> records with diff > 0.01: {(total_diff > 0.01).sum()}")

    # 3) Subtotal Check: Total Excluding Tax = Quantity * Unit Price
    calc_excl = df['Quantity'] * df['Unit Price']
    excl_diff = (calc_excl - df['Total Excluding Tax']).abs()
    print(f"3) Quantity x Unit Price check -> records with diff > 0.01: {(excl_diff > 0.01).sum()}")

    # 4) Item Breakdown Check: Total Dry Items + Total Chiller Items = Quantity
    items_diff = (df['Total Dry Items'] + df['Total Chiller Items'] - df['Quantity']).abs()
    print(f"4) Dry + Chiller = Quantity check -> records with diff: {(items_diff != 0).sum()}")

    # 5) Date Sequence Check: Delivery Date must be >= Invoice Date
    date_issue = (df['Delivery Date Key'] < df['Invoice Date Key']).sum()
    print(f"5) Delivery Date before Invoice Date -> records with issue: {date_issue}")

    # 6) Unregistered Customers Check: Count Customer Key == 0
    unknown_customers = (df['Customer Key'] == 0).sum()
    pct = (unknown_customers / len(df) * 100) if len(df) > 0 else 0
    print(f"6) Unregistered/Walk-in Customers (Key = 0): {unknown_customers} ({pct:.1f}%)")

    print('#' * 50)
    return df
    
df = data_validation(df)
```
